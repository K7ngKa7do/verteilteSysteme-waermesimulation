# C4 – Wärmeausbreitung auf der Metallplatte

Dieses Notebook zeigt die vier MPI-Ergebnisse auf dem Ubuntu-Manager nach 0, 500, 2000 und 8000 Zeitschritten. Führe die beiden Codezellen der Reihe nach aus.

Die Eingabedateien liegen in `/home/itmvs/waerme/niklas/`. Dort speichert das Notebook auch `waerme_verlauf.png`. Alle Bilder verwenden dieselbe Farbskala, damit dieselbe Temperatur in jedem Bild dieselbe Farbe hat.


```python
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
from matplotlib.colors import Normalize

datenordner = Path("/home/itmvs/waerme/niklas")
zeitschritte = (0, 500, 2000, 8000)
gitter_nach_schritten = {}

for schritte in zeitschritte:
    datei = datenordner / f"gitter_N128_P4_S{schritte}.npy"
    try:
        gitter = np.load(datei, allow_pickle=False)
    except FileNotFoundError as fehler:
        raise FileNotFoundError(f"MPI-Ergebnis fehlt: {datei}") from fehler

    if gitter.shape != (128, 128):
        raise ValueError(f"{datei.name}: Gitterform {gitter.shape}, erwartet (128, 128).")

    gitter_nach_schritten[schritte] = gitter
```

```python
fig, achsen = plt.subplots(1, 4, figsize=(14, 3.8), constrained_layout=True)
farbskala = Normalize(vmin=0, vmax=100)

for achse, schritte in zip(achsen, zeitschritte):
    bild = achse.imshow(
        gitter_nach_schritten[schritte],
        cmap="inferno",
        norm=farbskala,
        origin="upper",
        interpolation="nearest",
    )
    achse.set_title(f"Nach {schritte} Zeitschritten")
    achse.set_axis_off()

fig.colorbar(bild, ax=achsen, label="Temperatur", ticks=[0, 25, 50, 75, 100])
fig.savefig(datenordner / "waerme_verlauf.png", dpi=150)
plt.show()
```
